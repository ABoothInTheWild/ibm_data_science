### Step 1: Import Required Libraries


```python
import pandas as pd
```

### Step 2: Load the Dataset Into a Dataframe


#### **Read Data**
<p>
The functions below will download the dataset into your browser:
</p>


```python
file_path ="https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/UDKAZw-kz18Yj8P6icf_qw/survey-data-duplicates.csv"
df = pd.read_csv(file_path)

# Display the first few rows to ensure it loaded correctly
df.head()
```

### Step 3. Finding and Removing Duplicates
##### Task 1: Identify duplicate rows in the dataset.


```python
## Write your code here
# Identify duplicate rows in the dataset
duplicate_count = df.duplicated().sum()
print(f"Number of duplicate rows: {duplicate_count}")
df[df.duplicated(keep=False)].head()
```

##### Task 2: Remove the duplicate rows from the dataframe.



```python
## Write your code here
# Remove duplicate rows from the dataframe
print(f"Original shape: {df.shape}")
df = df.drop_duplicates()
print(f"Shape after removing duplicates: {df.shape}")
print(f"Remaining duplicates: {df.duplicated().sum()}")
```

### Step 4: Finding Missing Values
##### Task 3: Find the missing values for all columns.


```python
## Write your code here
# Find missing values for all columns
missing_values = df.isnull().sum().sort_values(ascending=False)
print(missing_values[missing_values > 0])
```

##### Task 4: Find out how many rows are missing in the column RemoteWork.


```python
## Write your code here
# Count missing rows in the RemoteWork column
missing_remotework = df['RemoteWork'].isnull().sum()
print(f"Number of missing rows in RemoteWork: {missing_remotework}")
```

### Step 5. Imputing Missing Values
##### Task 5: Find the value counts for the column RemoteWork.


```python
## Write your code here
# Find value counts for the RemoteWork column
print(df['RemoteWork'].value_counts())
```

##### Task 6: Identify the most frequent (majority) value in the RemoteWork column.



```python
## Write your code here
# Identify the most frequent value in the RemoteWork column
most_frequent_remotework = df['RemoteWork'].mode()[0]
print(f"Most frequent RemoteWork value: {most_frequent_remotework}")
```

##### Task 7: Impute (replace) all the empty rows in the column RemoteWork with the majority value.



```python
## Write your code here
# Impute missing values in RemoteWork with the most frequent value
df['RemoteWork'] = df['RemoteWork'].fillna(most_frequent_remotework)
print(f"Remaining missing values in RemoteWork: {df['RemoteWork'].isnull().sum()}")
```

##### Task 8: Check for any compensation-related columns and describe their distribution.



```python
## Write your code here
# Check compensation-related columns and describe their distribution
comp_columns = [col for col in df.columns if 'Comp' in col]
print("Compensation-related columns:")
print(comp_columns)

for col in comp_columns:
    print(f"\n{col} distribution:")
    print(df[col].describe())
```
