## Tasks


<h2>1. Load the Dataset</h2>
<p>
We use the <code>pandas.read_csv()</code> function for reading CSV files. However, in this version of the lab, which operates on JupyterLite, the dataset needs to be downloaded to the interface using the provided code below.

The functions below will download the dataset into your browser:



```python
# Define the URL of the dataset
file_path ="https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/UDKAZw-kz18Yj8P6icf_qw/survey-data-duplicates.csv"

# Load the dataset into a DataFrame
df = pd.read_csv(file_path)

# Display the first few rows to ensure it loaded correctly
df.head()
```

### 2. Explore the Dataset
##### Task 1: Display basic information and summary statistics of the dataset.


```python
## Write your code here
# Display basic information and summary statistics
print(df.info())
print("\nSummary statistics:")
df.describe()
```

### 3. Finding Missing Values
##### Task 2: Identify missing values for all columns.


```python
## Write your code here
# Identify missing values for all columns
missing_values = df.isnull().sum().sort_values(ascending=False).head(10)
print(missing_values)
```

##### Task 3: Visualize missing values using a heatmap (Using seaborn library).



```python
## Write your code here
# Visualize missing values using a heatmap
plt.figure(figsize=(12, 8))
sns.heatmap(df.isnull(), cbar=False, yticklabels=False)
plt.title('Missing Values Heatmap')
plt.show()
```

##### Task 4: Count the number of missing rows for a specific column (e.g., `EdLevel`).


```python
## Write your code here
# Count missing rows in the EdLevel column
missing_edlevel = df['EdLevel'].isnull().sum()
print(f"Number of missing rows in EdLevel: {missing_edlevel}")
```

### 4. Imputing Missing Values
##### Task 5: Identify the most frequent (majority) value in a specific column (e.g., `EdLevel`).


```python
## Write your code here
# Identify the most frequent value in the EdLevel column
most_frequent_edlevel = df['EdLevel'].mode()[0]
print(f"Most frequent EdLevel value: {most_frequent_edlevel}")
```

##### Task 6: Impute missing values in the `EdLevel` column with the most frequent value.



```python
## Write your code here
# Impute missing values in EdLevel with the most frequent value
df['EdLevel'] = df['EdLevel'].fillna(most_frequent_edlevel)
print(f"Remaining missing values in EdLevel: {df['EdLevel'].isnull().sum()}")
```

### 5. Visualizing Imputed Data
##### Task 7: Visualize the distribution of a column after imputation (e.g., `EdLevel`).


```python
## Write your code here
# Visualize the distribution of EdLevel after imputation
df['EdLevel'].value_counts().plot(kind='bar', figsize=(10, 5))
plt.title('Distribution of EdLevel after Imputation')
plt.xlabel('EdLevel')
plt.ylabel('Count')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```
