### Step 1: Import Required Libraries


```python
import pandas as pd
```

### Step 2: Load the Dataset into a DataFrame



load the dataset using pd.read_csv()


```python
# Define the URL of the dataset
file_path ="https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/UDKAZw-kz18Yj8P6icf_qw/survey-data-duplicates.csv"

# Load the dataset into a DataFrame
df = pd.read_csv(file_path)

# Display the first few rows to ensure it loaded correctly
print(df.head())
```

### Step 3: Identifying Duplicate Rows


**Task 1: Identify Duplicate Rows**
  1. Count the number of duplicate rows in the dataset.
  2. Display the first few duplicate rows to understand their structure.


```python
## Write your code here
# Count the number of duplicate rows
duplicate_count = df.duplicated().sum()
print(f"Number of duplicate rows: {duplicate_count}")
```

### Step 4: Removing Duplicate Rows


**Task 2: Remove Duplicates**
   1. Remove duplicate rows from the dataset using the drop_duplicates() function.
2. Verify the removal by counting the number of duplicate rows after removal .


```python
## Write your code here
# 1. Remove duplicate rows and keep the first occurrence
print(f"Original shape of the DataFrame: {df.shape}")
df.drop_duplicates(inplace=True)

# 2. Verify the removal
remaining_duplicates = df.duplicated().sum()
print(f"Number of duplicate rows after removal: {remaining_duplicates}")
print(f"New shape of the DataFrame: {df.shape}")
```

### Step 5: Handling Missing Values


**Task 3: Identify and Handle Missing Values**
   1. Identify missing values for all columns in the dataset.
   2. Choose a column with significant missing values (e.g., EdLevel) and impute with the most frequent value.


```python
## Write your code here
# 1. Identify missing values for all columns
missing_values = df.isnull().sum()
print("Missing values per column:")
print(missing_values[missing_values > 0]) # Displaying only columns with missing values

# 2. Impute 'EdLevel' with the most frequent value (mode)
most_frequent_edlevel = df['EdLevel'].mode()[0]
df['EdLevel'].fillna(most_frequent_edlevel, inplace=True)

print(f"\nMost frequent EdLevel used for imputation: {most_frequent_edlevel}")
print(f"Remaining missing values in EdLevel: {df['EdLevel'].isnull().sum()}")
```

### Step 6: Normalizing Compensation Data


**Task 4: Normalize Compensation Data Using ConvertedCompYearly**
   1. Use the ConvertedCompYearly column for compensation analysis as the normalized annual compensation is already provided.
   2. Check for missing values in ConvertedCompYearly and handle them if necessary.


```python
## Write your code here
# 1. Check for missing values in ConvertedCompYearly
missing_comp = df['ConvertedCompYearly'].isnull().sum()
print(f"Missing values in ConvertedCompYearly: {missing_comp}")

# 2. Handle missing values (Imputing with the median value to minimize outlier distortion)
median_comp = df['ConvertedCompYearly'].median()
df['ConvertedCompYearly'].fillna(median_comp, inplace=True)

print(f"Median ConvertedCompYearly used for imputation: {median_comp}")
print(f"Remaining missing values in ConvertedCompYearly: {df['ConvertedCompYearly'].isnull().sum()}")
```
