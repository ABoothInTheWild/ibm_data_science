## Extracting a seasonality feature
Now consider the `Date` column. We expect the weather patterns to be seasonal, having different predictablitiy levels in winter and summer for example.  
There may be some variation with `Year` as well, but we'll leave that out for now.
Let's engineer a `Season` feature from `Date` and drop `Date` afterward, since it is most likely less informative than season. 
An easy way to do this is to define a function that assigns seasons to given months, then use that function to transform the `Date` column.


### Create a function to map dates to seasons


````python
def date_to_season(date):
    month = date.month
    if (month == 12) or (month == 1) or (month == 2):
        return 'Summer'
    elif (month == 3) or (month == 4) or (month == 5):
        return 'Autumn'
    elif (month == 6) or (month == 7) or (month == 8):
        return 'Winter'
    elif (month == 9) or (month == 10) or (month == 11):
        return 'Spring'
````

### Define separate transformers for both feature types and combine them into a single preprocessing transformer


````python
# Scale the numeric features
numeric_transformer = Pipeline(steps=[('scaler', StandardScaler())])

# One-hot encode the categoricals 
categorical_transformer = Pipeline(steps=[('onehot', OneHotEncoder(handle_unknown='ignore'))])
````

## Baseline

Now let's extract the feature importances and plot them as a bar graph.


````python
# Combine numeric and categorical feature names
feature_names = numeric_features + list(grid_search.best_estimator_['preprocessor']
                                        .named_transformers_['cat']
                                        .named_steps['onehot']
                                        .get_feature_names_out(categorical_features))

feature_importances = grid_search.best_estimator_['classifier'].feature_importances_

importance_df = pd.DataFrame({'Feature': feature_names,
                              'Importance': feature_importances
                             }).sort_values(by='Importance', ascending=False)

N = 20  # Change this number to display more or fewer features
top_features = importance_df.head(N)

# Plotting
plt.figure(figsize=(10, 6))
plt.barh(top_features['Feature'], top_features['Importance'], color='skyblue')
plt.gca().invert_yaxis()  # Invert y-axis to show the most important feature on top
plt.title(f'Top {N} Most Important Features in predicting whether it will rain today')
plt.xlabel('Importance Score')
plt.show()
````

## Fixing class imbalances -- just SMOTE it!

**SMOTE (Synthetic Minority Over-sampling Technique)** is a method of over sampling the minority class, by generating synthetic examples rather than oversampling existing examples. It basically interpolates new data points between the existing points.

![](smote.png)

 If we just apply it outside of the pipeline, we can see the change in the distribution of the target class.

````python
from imblearn.over_sampling import SMOTE
smote = SMOTE(random_state=42)
X_train_resampled, y_train_resampled = smote.fit_resample(X_train[numeric_features], y_train)

plt.bar(y_train_resampled.value_counts().index, y_train_resampled.value_counts().values)
plt.xlabel('RainToday') 
plt.ylabel('Count')
plt.title('Distribution of RainToday after SMOTE Resampling')
plt.show()
````

We can integrate SMOTE into our pipeline, with our optimized logistic regression classifier, and see if we get better results.

````python
from imblearn.over_sampling import SMOTE
from imblearn.pipeline import Pipeline as ImbPipeline
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report, confusion_matrix
# Define the imbalanced pipeline
imb_pipeline = ImbPipeline(steps=[
    ('preprocessor', preprocessor),
    ('oversampler', SMOTE(random_state=42)),
    ('classifier', grid_search.best_estimator_.named_steps['classifier'])
])
imb_pipeline.fit(X_train, y_train)
y_pred_imb = imb_pipeline.predict(X_test)
print("Logistic Regression with SMOTE Performance:")
print(classification_report(y_test, y_pred_imb))
cm = confusion_matrix(y_test, y_pred_imb)
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', cbar=False)
plt.xlabel('Predicted')
plt.ylabel('Actual')
plt.title('Confusion Matrix for Logistic Regression with Oversampling')
plt.show()
````

That does seem to improve matters: we have gone from 51% to 74% on the Recall for the positive case. It is not perfect though, and we have tipped things towards more False Positives as well. 

Which do we prefer, more false positives or more false negatives? This can depend on our use case. For example, erring on the side of fals positives in medicine can lead to closer assessment rather than sending seriously ill patients home prematurely. On the other hand, spam detectors may move too many valid (and maybe important) emails to the spam folder and remain unread. 

Perhaps if there are more false positive predictions of rain we may just find ourselves unnecessarily carrying our umbrellas, but perhaps that is better than getting soaked on days we were promised it'd be sunny and hot.
